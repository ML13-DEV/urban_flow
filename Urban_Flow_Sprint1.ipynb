````python
# Arreglo de fechas inválidas y normalización a formato YYYY-MM-DD.

df['fecha'] = df['fecha'].replace("invalid_date", "1932-01-01")
df['fecha'] = to_datetime(df['fecha'], format='mixed',dayfirst=True).dt.strftime('%Y-%m-%d')

print(df.head())
````

````python
# Reemplazo de horas inválidas por la hora 00:00,
# eliminación de espacios extras y conversión a
# tipo datetime con formato H:M.
df['hora'] = df['hora'].replace("invalid_hour", "00:00")

df['hora'] = df['hora'].str.replace(
    r"[\s]",
    "",
    regex=True
)

df['hora'] = to_datetime(
    df['hora'],
    format='mixed'
).dt.strftime('%H:%M')

print(df.head(10))
````

````python
#Normalización de ubicaciones, eliminación de caracteres especiales y
# conversión a mayúsculas.

df['ubicacion'] = df['ubicacion'].replace(
    r'[^a-zA-Z0-9\s]',
    "" ,
    regex=True
).str.upper()

print(df['ubicacion'].head(10))
````

````python
# Limpieza y normalización de patentes, en caso de error se completa con NA
#  y conversión a mayúsculas.

df['patente'] = df['patente'].replace(
    r'[^a-zA-Z0-9]',
    "",
    regex=True
).str.upper()

df['patente'] = df['patente'].replace(
    r'[\s]',
    "",
    regex=True
)

df['patente'] = df['patente'].str.strip()
df['patente'] = df['patente'].fillna(NA)
df['patente'] = df['patente'].astype("string")

print(df['patente'].tail(10))
````

````python
# Limpieza de registros incompletos en columnas relevantes y cálculo de total de
# filas eliminadas.
columnas_relevantes: list[str] = [
    'patente',
    'fecha',
    'velocidad_registrada',
    'velocidad_maxima'
]

filas_antes: int = df.shape[0]

df = df.dropna(subset=columnas_relevantes)

filas_despues: int = df.shape[0]
filas_eliminadas: int = filas_antes - filas_despues

print(f"Se eliminaron {filas_eliminadas} filas.")
print("Las columnas observadas son:")

for columna in columnas_relevantes:
    print(f"- {columna}")
````

````python
# Detección de outliers mediante Rango Intercuartílico

Q1: float = df['velocidad_registrada'].quantile(0.25)
Q3: float = df['velocidad_registrada'].quantile(0.75)
IQR: float = Q3 - Q1

limite_inferior: float = Q1 - 1.5 * IQR
limite_superior: float = Q3 + 1.5 * IQR

outliers = df[
    (df['velocidad_registrada'] < limite_inferior)
    | (df['velocidad_registrada'] > limite_superior)
]

min_valor: float = df['velocidad_registrada'].min()
max_valor: float = df['velocidad_registrada'].max()

print("Estadísticas de 'velocidad_registrada':")
print(f"Q1: {Q1} | Q3: {Q3} | IQR: {IQR}")
print(f"Límites: [{limite_inferior} , {limite_superior}]")
print(f"Rango real: [{min_valor} , {max_valor}]")
print(f"Cantidad de outliers detectados: {len(outliers)}")
````

````python
# Creación de la columna 'exceso_velocidad_real' mediante el calculo de
# velocidad_registrada menos velocidad_maxima
df['exceso_velocidad_real'] = (
    df['velocidad_registrada']
    - df['velocidad_maxima']
)

print(
    df[['patente', 'exceso_velocidad_real']]
    .head(10)
)
````

````python
# Creación de la columna 'exceso_velocidad' mediante el calculo de
# exceso_velocidad_real más el producto de velocidad_maxima y 5%.
df['exceso_velocidad'] = (
    df['exceso_velocidad_real']
    + (df['velocidad_maxima'] * 0.05)
)

print(
    df[['patente', 'exceso_velocidad']]
    .head(10)
)
````

````python
# Cálculo de filas antes de la limpieza, eliminación de filas
# con exceso <= 0 y observación la cantidad e filas eliminadas.
filas_antes: int = df.shape[0]

indices_a_eliminar = df[
    df['exceso_velocidad'] <= 0
].index

df.drop(indices_a_eliminar, inplace=True)

filas_despues: int = df.shape[0]
filas_eliminadas: int = filas_antes - filas_despues

print(f"Se eliminaron {filas_eliminadas} filas")
````

## Ejercicio 04

> Puntos: 2

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

- Reciba el DataFrame limpio como argumento de inicialización.

- Encapsule correctamente los datos.

- Implemente métodos para:

  - Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

  - Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.


  - Exceso promedio de velocidad. Debe retorna un flotante.

  - Exceso real promedio de velocidad . Debe retorna un flotante.

  - Contabilizar la cantidad de multas por cada  ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

- Crear el objecto e invocar cada uno de los métodos en celdas separadas.
  

````python
# Creación de clase FineAnalyzer con sus métodos

class FineAnalyzer:
    """
    Clase para el análisis de infracciones de tránsito (Ejercicio 04).
    Encapsula rankings, cálculos de promedios y conteos por ubicación.
    """

    def __init__(self, df: DataFrame) -> None:
        """
        Inicializa la clase con una copia interna del DataFrame.
        """
        self._df: DataFrame = df.copy()

    def __str__(self) -> str:
        return (
            f"FineAnalyzer cargado con {len(self._df)} registros."
        )

    def mas_multadas(self) -> DataFrame:
        """
        Retorna un ranking de las 5 patentes con más infracciones.
        """
        top: Series = self._df.groupby(
            'patente'
        )['multa_id'].count()

        top = top.sort_values(ascending=False).head(5)

        return DataFrame(
            {
                "patente": top.index,
                "cantidad": top.values
            },
            index=range(1, len(top) + 1)
        )

    def top_horarios(self) -> DataFrame:
        """
        Retorna las 5 franjas horarias con mayor cantidad de multas.
        """
        horas_dt: Series = to_datetime(
            self._df['hora'],
            format='%H:%M'
        ).dt.hour

        top_h: Series = self._df.groupby(
            horas_dt
        )['multa_id'].count()

        top_h = top_h.sort_values(ascending=False).head(5)

        return DataFrame(
            {
                "hora": top_h.index,
                "cantidad": top_h.values
            },
            index=range(1, len(top_h) + 1)
        )

    def promedio_exceso(self) -> float:
        """
        Calcula el exceso promedio de velocidad considerando
        la tolerancia.
        """
        return float(
            np.round(self._df['exceso_velocidad'].mean(), 2)
        )

    def promedio_exceso_real(self) -> float:
        """
        Calcula el exceso real promedio de velocidad
        (sin tolerancia).
        """
        return float(
            np.round(self._df['exceso_velocidad_real'].mean(), 2)
        )

    def multas_ubicacion(self) -> DataFrame:
        """
        Retorna la cantidad de multas agrupadas por ubicación
        geográfica.
        """
        ubis: Series = self._df.groupby(
            'ubicacion'
        )['multa_id'].count()

        ubis = ubis.sort_index()

        return DataFrame(
            {
                "ubicacion": ubis.index,
                "cantidad": ubis.values
            },
            index=range(1, len(ubis) + 1)
        )
````

````python
# Gráfico: Top 10 Patentes más multadas
def ej5_0() -> None:
    """
    Genera y guarda un gráfico de barras con el Top 10 de las
    patentes más multadas. La imagen se exporta en alta resolución
    en la carpeta de resultados.
    """

    df_grafico: Series = df_interim.groupby(
        by='patente'
    )['multa_id'].count()

    df_grafico = df_grafico.sort_values(
        ascending=False
    ).head(10)

    plt.figure(figsize=(10, 5))

    df_grafico.plot(
        kind='bar',
        xlabel='Patentes',
        ylabel='Multas',
        title='Patentes más multadas'
    )

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.grid(True)

    plt.savefig(
        "/content/urban_flow/data/interim/plots/patentes.jpg"
    )

    plt.show()


ej5_0()
````

````python
# Gráfico: Porcentajes de infracciones por horas
def ej5_1() -> None:
    """
    Genera un gráfico de torta con la distribución porcentual
    de infracciones por hora. Exporta el resultado a la carpeta
    de gráficos.
    """

    df_interim['hora'] = to_datetime(
        df_interim['hora'],
        format='%H:%M'
    )

    df_horarios: Series = df_interim.groupby(
        by=df_interim['hora'].dt.hour
    )['multa_id'].count()

    df_horarios = df_horarios.sort_values(
        ascending=False
    )

    plt.figure(figsize=(8, 8))

    df_horarios.plot(
        kind='pie',
        title='Multas por hora',
        ylabel='',
        colormap='coolwarm',
        autopct='%.1f%%',
        startangle=90,
        pctdistance=0.85
    )

    plt.grid(True)

    plt.savefig(
        "/content/urban_flow/data/interim/plots/hours.jpg"
    )

    plt.show()


ej5_1()
````

````python
# Gráfico: Cantidad de multas por mes
def ej5_2() -> None:
    """
    Genera un gráfico de barras con la cantidad de multas por mes,
    ordenado de mayor a menor frecuencia.
    """
    df_interim['fecha'] = to_datetime(
        df_interim['fecha'],
        format='%Y-%m-%d'
    )

    df_grafico: Series = df_interim.groupby(
        df_interim['fecha'].dt.month
    )['multa_id'].count()

    df_grafico = df_grafico.sort_values(
        ascending=False
    )

    plt.figure(figsize=(10, 5))

    df_grafico.plot(
        kind='bar',
        xlabel='Meses',
        ylabel='Multas',
        title='Cantidad de multas por mes'
    )

    plt.grid(True)
    plt.xticks(rotation=0)

    plt.savefig(
        "/content/urban_flow/data/interim/plots/months.jpg"
    )

    plt.show()

ej5_2()
````

````python
# Gráfico: Excesos de velocidad a la hora 00:00
def ej5_3() -> None:
    """
    Genera un gráfico de líneas que muestra la distribución
    de los excesos de velocidad para la hora 00:00.
    """

    df_interim['hora'] = to_datetime(
        df_interim['hora'],
        format='%H:%M'
    )

    df_solo_medianoche: DataFrame = df_interim[
        df_interim['hora'].dt.hour == 0
    ]

    df_grafico: Series = df_solo_medianoche.groupby(
        'exceso_velocidad'
    )['multa_id'].count()

    plt.figure(figsize=(10, 5))

    df_grafico.plot(
        kind='line',
        title=(
            'Excesos de velocidad agrupados en la hora 00:00'
        )
    )

    plt.xlabel("Velocidades")
    plt.ylabel("Excesos")
    plt.grid(True)

    plt.savefig(
        "/content/urban_flow/data/interim/plots/hour.jpg"
    )

    plt.show()

ej5_3()
````

````python
# Gráfico: Excesos de velocidad en la fecha 1932-01-01
def ej5_4() -> None:
    """
    Genera un gráfico de líneas que muestra la distribución de
    los excesos de velocidad para la fecha 1932-01-01.
    """

    df_interim['fecha'] = to_datetime(
        df_interim['fecha'],
        format='%Y-%m-%d'
    )

    df_fecha_especifica: DataFrame = df_interim[
        df_interim['fecha'] == '1932-01-01'
    ]

    df_grafico: Series = df_fecha_especifica.groupby(
        'exceso_velocidad'
    )['multa_id'].count()

    plt.figure(figsize=(10, 5))

    df_grafico.plot(
        kind='line',
        title=(
            'Excesos de velocidad agrupados en la fecha '
            '1932-01-01'
        )
    )

    plt.xlabel("Velocidades")
    plt.ylabel("Excesos")
    plt.grid(True)

    plt.savefig(
        "/content/urban_flow/data/interim/plots/date.jpg"
    )

    plt.show()

ej5_4()
````
