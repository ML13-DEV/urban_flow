El diccionario es (guardarlo en `ubran_flow/data/interim/group_images.json`):

group_images = {
  'plates': [{

    'area': cálculo del área de la imagen,

- Desarrollar un función para mostrar 8 imágenes de forma aleatorea en forma de tabla de 2 columnas y 4 filas. Debe mostrar 4 imágenes de cada grupo. Es una función reutilizable. En este punto se deben mostrar las imágenes originales. Cada imágen debe tener un ancho de 10 pulgas y un alto de 8 pulgadas, se debe mostrar el nombre de la imagen.



````python
#@title Clasificación de las patentes según la foto

PATH_IMGS = "data/raw/imgs/urban_flow_plates/"

group_images = {'plates':[], 'completes':[]}

print(f"{'Archivo':<40} | {'Tamaño':<10}")
print("-" * 70)

for archivo in os.listdir(PATH_IMGS):
    ruta_completa = os.path.join(PATH_IMGS, archivo)

    if os.path.isfile(ruta_completa):
        img = cv2.imread(ruta_completa)

        if img is not None:
            alto, ancho, _ = img.shape
            resolucion = f"{ancho}x{alto}"

            tamanio_bytes = os.path.getsize(ruta_completa)
            tamanio_kb = tamanio_bytes / 1024

            if 'patente' in archivo:
              group_images['plates'].append({
                    'filename': archivo,
                    'width': ancho,
                    'height': alto,
                    'area': ancho * alto,
                    'path': ruta_completa,
                    'patent': ''
                })
            else:
                group_images['completes'].append({
                    'filename': archivo,
                    'width': ancho,
                    'height': alto,
                    'area': ancho * alto,
                    'path': ruta_completa,
                    'patent': ''
                })

            print(f"{archivo:<40} | {tamanio_kb:>7.2f} KB")
````

````python
#@title Cantidad de patentes por grupo
patentes = len(group_images['plates'])
completes = len(group_images['completes'])
print(f"El grupo 'plates' contiene {patentes} imágenes.")
print(f"El grupo 'completes' contiene {completes} imágenes.")
````

````python
#@title Creamos la función para mostrar 8 imágenes al azar

def mostrar_reporte_imagenes(grupo_imagenes, carpeta_base=""):
    """
    Muestra una grilla de 4x2 con imágenes seleccionadas aleatoriamente de dos grupos:
    'plates' y 'completes'. La función puede leer las imágenes desde las rutas
    especificadas en `grupo_imagenes` o, si se proporciona `carpeta_base`,
    desde subcarpetas dentro de esa base (e.g., 'carpeta_base/plates').
    Cada imagen se visualiza en un tamaño de 10x8 pulgadas y su nombre se muestra
    como título.

    Parámetros:
    -----------
    grupo_imagenes : dict
        Diccionario que contiene listas de metadatos de imágenes para los grupos
        'plates' y 'completes'. Cada elemento de la lista es un diccionario
        con al menos la clave 'path' que indica la ruta completa de la imagen.
    carpeta_base : str, opcional
        Si se proporciona, la función buscará las imágenes en las subcarpetas
        'plates' y 'completes' dentro de esta ruta. Por defecto, usa las rutas
        directamente de `grupo_imagenes`.

    Retorna:
    --------
    None
        La función no retorna ningún valor; solo muestra una figura de Matplotlib.
    """

    rutas_patentes: list = []
    rutas_completes: list = []

    if carpeta_base != "":
      directorio_patentes = os.path.join(carpeta_base, 'plates')
      directorio_completes = os.path.join(carpeta_base, 'completes')
      # Tomamos las muestras al azar del diccionario original
      for archivo in os.listdir(directorio_patentes):
        if archivo.endswith(('.png', '.jpg', '.jpeg', '.qoi', '.jp2', '.PNG', '.JPG')):
          rutas_patentes.append(os.path.join(directorio_patentes, archivo))

      for archivo in os.listdir(directorio_completes):
        if archivo.endswith(('.png', '.jpg', '.jpeg', '.qoi', '.jp2', '.PNG', '.JPG')):
          rutas_completes.append(os.path.join(directorio_completes, archivo))
    else:
      rutas_patentes = [img['path'] for img in grupo_imagenes['plates']]
      rutas_completes = [img['path'] for img in grupo_imagenes['completes']]

    muestra_patentes = random.sample(rutas_patentes, 4)
    muestra_completas = random.sample(rutas_completes, 4)


    fig, axes = plt.subplots(4, 2, figsize=(20, 32))
    fig.subplots_adjust(hspace=0.2, wspace=0.1)

    for i in range(4):
        datos_p = muestra_patentes[i]
        datos_c = muestra_completas[i]

        nombre_p = os.path.basename(datos_p)
        nombre_c = os.path.basename(datos_c)

        if 'qoi' in datos_p.lower():
            img_p = qoi.read(datos_p)
        else:
            img_p = cv2.imread(datos_p, cv2.IMREAD_UNCHANGED)

        if img_p is not None:
            axes[i, 0].imshow(img_p, cmap='gray' if len(img_p.shape) == 2 else None)
        axes[i, 0].set_title(f"Plate: {datos_p}", fontsize=14, color='dimgray', pad=10)
        axes[i, 0].axis('off')

        img_c = cv2.imread(datos_c, cv2.IMREAD_UNCHANGED)

        if img_c is not None:
            axes[i, 1].imshow(img_c, cmap='gray' if len(img_c.shape) == 2 else None)
        axes[i, 1].set_title(f"Complete: {datos_c}", fontsize=14, color='dimgray', pad=10)
        axes[i, 1].axis('off')

    plt.tight_layout()
    plt.show()
````

````python
#@title Probamos la función pasándole la lista de diccionarios con las patentes clasificadas previamente
mostrar_reporte_imagenes(group_images)
````

````python
#@title Suavizado de las imágenes en escala de grises y almacenado

os.makedirs('data/interim/imgs/03_02_blur/plates', exist_ok=True)
os.makedirs('data/interim/imgs/03_02_blur/completes', exist_ok=True)

for img in os.listdir('data/interim/imgs/03_01_gray_scale/plates/'):
  if img.endswith(extensiones_validas):
    archivo = cv2.imread(f'data/interim/imgs/03_01_gray_scale/plates/{img}')

    if archivo is not None:
      img_blur = cv2.GaussianBlur(archivo,(7,7), 0)

      cv2.imwrite(f'data/interim/imgs/03_02_blur/plates/{img}', img_blur)

for img in os.listdir('data/interim/imgs/03_01_gray_scale/completes/'):
  if img.endswith(extensiones_validas):
    archivo = cv2.imread(f'data/interim/imgs/03_01_gray_scale/completes/{img}')

    if archivo is not None:
      img_blur = cv2.GaussianBlur(archivo,(7,7), 0)

      cv2.imwrite(f'data/interim/imgs/03_02_blur/completes/{img}', img_blur)
````

````python
#@title Definición de la función extraer_patente

import easyocr
def extraer_patente(image_path) -> str:
  """ Extrae la patente de una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Patente extraída.
  """
  reader = easyocr.Reader(['es'])
  detection:list = reader.readtext(image_path)
  if len(detection) == 0:
    return ''
  patent = ''
  for d in detection:
    if isinstance(d,tuple) and len(d) == 3:
      if isinstance(d[1],str):
        patent+=d[1]
  return patent
````

````python
#@title Definición de la función 'vincular_imagenes_multas'

def vincular_imagenes_multas(lista_imagenes: list,
                             ruta_csv_interim: str,
                             ruta_csv_processed: str) -> pd.DataFrame:
    """
    Procesa las imágenes originales a color desde la lista de imágenes
    clasificadas y las vincula directamente con el dataset de multas
    utilizando EasyOCR.

    Parámetros:
    -----------
    lista_imagenes : list
        Lista de diccionarios con los metadatos de las imágenes
        (cada elemento contiene 'filename' y 'path').
    ruta_csv_interim : str
        Ruta física del archivo CSV intermedio con el listado de multas.
    ruta_csv_processed : str
        Ruta física donde se exportará el archivo CSV final vinculado.

    Devuelve:
    ---------
    pd.DataFrame
        El DataFrame de multas actualizado con los mejores matches y las
        columnas nuevas.
    """

    df_multas: pd.DataFrame = pd.read_csv(ruta_csv_interim)
    df_multas['imagen'] = None
    df_multas['patente_imagen'] = None
    df_multas['ratio'] = 0.0


    img_data: dict
    for img_data in lista_imagenes:
        ruta_original: str = img_data['path']

        img_bgr: np.ndarray = cv2.imread(ruta_original)
        if img_bgr is None:
            continue


        alto, ancho = img_bgr.shape[:2]
        img_resizada: np.ndarray = cv2.resize(
            img_bgr,
             (ancho * 2, alto * 2),
            interpolation=cv2.INTER_CUBIC
        )

        img_rgb: np.ndarray = cv2.cvtColor(
            img_resizada,
            cv2.COLOR_BGR2RGB
        )

        texto_raw: str = extraer_patente(
            img_rgb
        ).strip().upper()

        texto_ocr: str = re.sub(
            r'[^A-Z0-9]',
            '',
            texto_raw
        )

        img_data['patent'] = texto_ocr if texto_ocr else ''

        if not texto_ocr:
            continue

        idx: int
        fila: pd.Series
        for idx, fila in df_multas.iterrows():
            patente_csv: str = str(
                fila['patente']
            ).strip().upper()

            # Calcular el porcentaje de coincidencia secuencial
            matcher: SequenceMatcher = SequenceMatcher(
                None,
                patente_csv,
                texto_ocr
            )

            ratio_actual: float = matcher.ratio()
            porcentaje_actual: float = round(
                ratio_actual * 100,
                2
            )

            ratio_previo: float = float(
                df_multas.at[
                    idx,
                    'ratio'
                    ]
                )

            if ratio_actual >= 0.80 and porcentaje_actual > ratio_previo:
                df_multas.loc[idx, 'imagen'] = ruta_original
                df_multas.loc[idx, 'patente_imagen'] = texto_ocr
                df_multas.loc[idx, 'ratio'] = porcentaje_actual

    os.makedirs(
        os.path.dirname(
            ruta_csv_processed),
         exist_ok=True
    )

    df_multas.to_csv(
        ruta_csv_processed,
        index=False
    )

    return df_multas
````
